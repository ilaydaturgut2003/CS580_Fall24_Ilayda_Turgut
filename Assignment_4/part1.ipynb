```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

data = pd.read_csv("data.csv", header=None)
data.columns = ['x1', 'x2', 'label']
X = data[['x1','x2']].values
y = data['label'].values

plt.scatter(X[:, 0], X[:, 1], c=y, cmap='viridis')
plt.xlabel('x1')
plt.ylabel('x2')
plt.title('Data Points')
plt.show()

iterations = 65 
learning_rate = 1

# Part 1
np.random.seed(42)  
weights = np.random.rand(2) 
bias = np.random.rand(1)

def update_weights_heuristic(X, y, weights, bias, learning_rate):
    for i in range(len(X)):
        prediction = np.dot(X[i], weights) + bias
        predicted_label = 1 if prediction >= 0 else 0
        
        if predicted_label != y[i]:
            error = y[i] - predicted_label
            bias += learning_rate * error
            weights += learning_rate * error * X[i]
    return weights, bias

plt.scatter(X[:, 0], X[:, 1], c=y, cmap='viridis')

for iteration in range(iterations):  
    slope = -weights[0] / weights[1]
    intercept = -bias / weights[1]
    x_vals = np.array([min(X[:, 0]), max(X[:, 0])])
    y_vals = slope * x_vals + intercept
    plt.plot(x_vals, y_vals, 'g--' if iteration < iterations - 1 else 'k-', alpha=0.7)  
    
    weights, bias = update_weights_heuristic(X, y, weights, bias, learning_rate)


plt.ylim(0, 1)
plt.yticks(np.arange(0, 1.1, 0.2))

plt.xlabel('x1')
plt.ylabel('x2')
plt.title('Solution Boundary')
plt.show()



```
